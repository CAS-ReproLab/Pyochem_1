```python
enzyme_data= {
    "Enzyme Name":["Amalase", "Lipase", "pepsin", "catalase", "DNA polymerase"],
    "Optimal pH": [7.0,8.0,2.0,7.0,9.0],
    "Optimal Temperature":[37,37,37,30,72]
}

print(enzyme_data)
```

```python
import pandas as pd

enzymes_df=pd.DataFrame(enzyme_data)
print(enzymes_df)
```

```python
# selecting only specfic columns 
enzymes_temp= enzymes_df[["Enzyme Name", "Optimal Temperature"]]
print(enzymes_temp)
```

```python
#Filtering for a PH > 7 

optimal_pH= enzymes_df[enzymes_df["Optimal pH"]>7]
print(optimal_pH)
```

```python
# adding a new ccolumn 

enzymes_df["thermophilic"]=enzymes_df["Optimal Temperature"]> 60 
print(enzymes_df)
```

```python
#creating a secon data frame with defciencey info 
inhibition_df= pd.DataFrame({
    "Enzyme Name":["Amalase", "Lipase", "pepsin", "catalase"],
    "Inhibition(%)": [12.5,40.0, None, 8.0]
})


# MERGE
merged_enzymes= pd.merge(enzymes_df, inhibition_df, on="Enzyme Name",how="left")
print(merged_enzymes)
                          
```

```python
#calculate the mean to fill in missing data

mean_inhib=merged_enzymes["Inhibition(%)"].mean()
print("Mean value:", mean_inhib)

merged_enzymes["Inhibition(%)"]= merged_enzymes["Inhibition(%)"]. fillna(mean_inhib)
print(merged_enzymes)
```

```python
import pandas as pd

vitamin_data = {
    "Vitamin": ["A", "C", "D", "B12"],
    "Daily_mg": [0.9, 90, 0.015, 0.0024],
    "Solubility": ["Fat", "Water", "Fat", "Water"]
}

vitamins_df = pd.DataFrame(vitamin_data)

# Add a new column based on a condition
vitamins_df["High_Dose"] = vitamins_df["Daily_mg"] > 1
print(vitamins_df)
```

```python


# First DataFrame
vitamins_df = pd.DataFrame({
    "Vitamin": ["A", "C", "D", "B12"],
    "Daily_mg": [0.9, 90, 0.015, 0.0024]
})

# Second DataFrame with deficiency info
deficiency_df = pd.DataFrame({
    "Vitamin": ["C", "D", "K"],
    "Deficiency": ["Scurvy", "Bone disorders", "Bleeding issues"]
})

# Merge the two on the "Vitamin" column
merged_vitamins = pd.merge(vitamins_df, deficiency_df, on="Vitamin", how="left")
print(merged_vitamins)

```
