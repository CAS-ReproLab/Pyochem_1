```python
df = pd.read_csv("Provisional_COVID-19_Death_Counts_by_Week_Ending_Date_and_State_20251118.csv")
df.head()
```

```python
## Converting the data column 

df["Week Ending Date"] = pd.to_datetime(df["Week Ending Date"])
```

```python
## Filter for a single state 
nc = df[df["State"] == "North Carolina"].copy()
nc.head()
```

```python
## BASIC STATS looking at just one state

nc["COVID-19 Deaths"].describe()
```

```python
## Pandas stored the column as objects (strings) instead of numbers.

nc["COVID-19 Deaths"] = pd.to_numeric(nc["COVID-19 Deaths"], errors="coerce")
```

```python
## Making a Stats table for an individual state
# Calculate individual stats
total_deaths = nc["COVID-19 Deaths"].sum()
mean_deaths = nc["COVID-19 Deaths"].mean()
median_deaths = nc["COVID-19 Deaths"].median()
std_deaths = nc["COVID-19 Deaths"].std()
min_deaths = nc["COVID-19 Deaths"].min()
max_deaths = nc["COVID-19 Deaths"].max()

# Identify rows for peak & minimum weeks
peak_week = nc.loc[nc["COVID-19 Deaths"].idxmax()]
min_week_row = nc.loc[nc["COVID-19 Deaths"].idxmin()]

# Create a clean stats table
stats_table = pd.DataFrame({
    "Metric": [
        "Total Deaths",
        "Average Weekly Deaths",
        "Median Weekly Deaths",
        "Standard Deviation",
        "Minimum Weekly Deaths",
        "Maximum Weekly Deaths",
        "Week of Minimum Deaths",
        "Week of Maximum Deaths",
        "Number of Weeks Reported"
    ],
    "Value": [
        total_deaths,
        mean_deaths,
        median_deaths,
        std_deaths,
        min_deaths,
        max_deaths,
        min_week_row["Week Ending Date"],
        peak_week["Week Ending Date"],
        nc["COVID-19 Deaths"].count()
    ]
})

stats_table
```

## When were national and local peaks? 
## (comparing national average to one state)

```python
## you have to install mat plot lib...Make sure to install through your enviornment install through your enviorment 
import matplotlib.pyplot as plt


## For North carolina
# Convert dates
df["Week Ending Date"] = pd.to_datetime(df["Week Ending Date"])

# Filter NC
nc = df[df["State"] == "North Carolina"].copy()

# Convert to numeric
nc["COVID-19 Deaths"] = pd.to_numeric(nc["COVID-19 Deaths"], errors="coerce")
```

```python
## Compute National average deaths per week

# Convert to numeric for entire dataset
df["COVID-19 Deaths"] = pd.to_numeric(df["COVID-19 Deaths"], errors="coerce")

# Group entire U.S. by week and compute average deaths across states
national_avg = (
    df.groupby("Week Ending Date")["COVID-19 Deaths"]
      .mean()
      .reset_index(name="National Average Deaths")
)
```

```python
## Merge NC data with national averages
nc_vs_nat = pd.merge(
    nc[["Week Ending Date", "COVID-19 Deaths"]],
    national_avg,
    on="Week Ending Date",
    how="inner"
)

## do this to confirm merge worked
nc_vs_nat.head()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    nc_vs_nat["Week Ending Date"],
    nc_vs_nat["COVID-19 Deaths"],
    label="North Carolina COVID-19 Deaths",
    linewidth=2
)

plt.plot(
    nc_vs_nat["Week Ending Date"],
    nc_vs_nat["National Average Deaths"],
    label="National Average COVID-19 Deaths",
    linewidth=2
)

plt.title("COVID-19 Weekly Deaths: North Carolina vs. National Average")
plt.xlabel("Week Ending Date")
plt.ylabel("Deaths")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

plt.show()

```

## Did certain regions experince earlier waves?

```python
## Create a region map

region_map = {
    # Northeast
    "Maine": "Northeast", "New Hampshire": "Northeast", "Vermont": "Northeast",
    "Massachusetts": "Northeast", "Rhode Island": "Northeast", "Connecticut": "Northeast",
    "New York": "Northeast", "New Jersey": "Northeast", "Pennsylvania": "Northeast",
    
    # Midwest
    "Ohio": "Midwest", "Indiana": "Midwest", "Illinois": "Midwest", 
    "Michigan": "Midwest", "Wisconsin": "Midwest", "Minnesota": "Midwest",
    "Iowa": "Midwest", "Missouri": "Midwest", "North Dakota": "Midwest",
    "South Dakota": "Midwest", "Nebraska": "Midwest", "Kansas": "Midwest",
    
    # South
    "Delaware": "South", "Maryland": "South", "District of Columbia": "South",
    "Virginia": "South", "West Virginia": "South", "North Carolina": "South",
    "South Carolina": "South", "Georgia": "South", "Florida": "South",
    "Kentucky": "South", "Tennessee": "South", "Alabama": "South",
    "Mississippi": "South", "Arkansas": "South", "Louisiana": "South",
    "Texas": "South", "Oklahoma": "South",
    
    # West
    "Montana": "West", "Idaho": "West", "Wyoming": "West", "Colorado": "West",
    "New Mexico": "West", "Arizona": "West", "Utah": "West", "Nevada": "West",
    "Washington": "West", "Oregon": "West", "California": "West",
    "Alaska": "West", "Hawaii": "West"
}

## Assign each state to its region 

df["Region"] = df["State"].map(region_map)
```

```python
## Compute weekly deaths per region 

region_weekly = (
    df.dropna(subset=["Region"])
      .groupby(["Region", "Week Ending Date"])["COVID-19 Deaths"]
      .sum()
      .reset_index()
)


```

```python
## Find peak week for each reason 

region_peaks = (
    region_weekly
    .loc[region_weekly.groupby("Region")["COVID-19 Deaths"].idxmax()]
    .sort_values("Week Ending Date")
)

region_peaks
```

```python
## Plot weekly deaths for all regions

plt.figure(figsize=(13,7))

for region in region_weekly["Region"].unique():
    sub = region_weekly[region_weekly["Region"] == region]
    plt.plot(sub["Week Ending Date"], sub["COVID-19 Deaths"], label=region, linewidth=2)

plt.title("Weekly COVID-19 Deaths by Census Region")
plt.xlabel("Week Ending Date")
plt.ylabel("Deaths")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Which states had the worst outbreaks?

```python
## Compute total deaths by state 

state_totals = (
    df.groupby("State")["COVID-19 Deaths"]
      .sum()
      .sort_values(ascending=False)
)

state_totals.head(10)
```

```python
##Make a bar graph

plt.figure(figsize=(12,6))
plt.bar(state_totals.head(10).index, state_totals.head(10).values)
plt.xticks(rotation=45, ha="right")
plt.title("Top 10 States by Total COVID-19 Deaths")
plt.ylabel("Total Deaths")
plt.tight_layout()
plt.show()
```

```python
##Peak weekly deaths per state 

# Ensure date is datetime
df["Week Ending Date"] = pd.to_datetime(df["Week Ending Date"], errors="coerce")

# Clean subset: real states only, valid date & deaths
clean = df[
    (df["State"] != "United States") &
    df["Week Ending Date"].notna() &
    df["COVID-19 Deaths"].notna()
].copy()

# Make sure deaths are numeric
clean["COVID-19 Deaths"] = pd.to_numeric(clean["COVID-19 Deaths"], errors="coerce")

# Drop any rows where conversion failed
clean = clean.dropna(subset=["COVID-19 Deaths"])

# Compute peak week per state
state_peaks = (
    clean.loc[clean.groupby("State")["COVID-19 Deaths"].idxmax()]
         [["State", "Week Ending Date", "COVID-19 Deaths"]]
         .sort_values("COVID-19 Deaths", ascending=False)
)

state_peaks.head(10)
```

```python
## Average weekly deaths per state 
state_avg = (
   ## remove the united states from table using the clean function 
    clean[clean["State"] != "United States"]
      .groupby("State")["COVID-19 Deaths"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

state_avg

```

```python
## Combining total deaths weekly deaths and average weekly deaths to give definition of worst 

# Combine: total, peak, avg for each state
summary = pd.DataFrame({
    "Total Deaths": state_totals,
    "Peak Weekly Deaths": state_peaks.set_index("State")["COVID-19 Deaths"],
    "Average Weekly Deaths": state_avg
})

# Sort by total deaths (or whichever metric you want)
summary_sorted = summary.sort_values("Total Deaths", ascending=False)

summary_sorted.head(10)
```
